# AI Data Exploration Project

**Link to Dataset:** https://www.kaggle.com/datasets/katerynameleshenko/ai-index

```python
import pandas as pd
df = pd.read_csv("AI_index_db.csv")
df
```

```python
total_score = df.sort_values(['Total score'], ascending = True)
total_score
```

```python
total_score.head()  # Countries with the lowest total score use least amount of AI. They all belong to the lower middle income groups.
```

```python
total_score.tail() # Countries with the highest total score use the most amount of AI. They all belong to upper middle and high income groups.
```

```python
# Create bar graphs to show the top 5 countries with the lowest and highest total scores.
# Create scatter plots to show the relationship between each co-variable and the total score to predict if using AI improves the efficiency of these co-variables.
```

```python
# Data Visualization.
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(nrows = 1, ncols = 2, sharex = False, sharey = True)

ax[0].bar(total_score.head().Country, total_score.head()['Total score']) # This creates the first bar graph.
ax[0].set_title("Countries that use the least AI")
ax[0].set_xlabel("Countries")
ax[0].set_ylabel("AI Total Score")
ax[0].tick_params(axis = 'x', rotation = 70)

ax[1].bar(total_score.tail().Country, total_score.tail()['Total score']) # This creates the second bar graph.
ax[1].set_title("Countries that use the most AI")
ax[1].set_xlabel("Countries")
ax[1].tick_params(axis = 'x', rotation = 70)

plt.tight_layout()
plt.show()
```

```python
plt.scatter(df['Talent'], df['Total score'])
plt.title("Correlation of Talent and Total Score")
plt.xlabel("Talent")    # Strong positive correlation: The more skilled practitioners for the provision of artificial intelligence solutions, the higher the total score in any country.
plt.ylabel("Total Score")

plt.tight_layout()
plt.show()
```

```python
correlation_coefficient = df['Talent'].corr(df['Total score']) # Strong positive correlation. Talent impacts the total score most because it can help a country achieve Long-Run Economic Growth.
print(f"Correlation Coefficient: {correlation_coefficient}")
```

```python
plt.scatter(df['Infrastructure'], df['Total score'])
plt.title("Correlation of Infrastructure and Total Score")
plt.xlabel("Infrastructure")  # Positive correlation: The more reliability and scale of access infrastructure, from electricity and internet, to super computing capabilities, the higher the total score in any country.
plt.ylabel("Total Score")

plt.tight_layout()
plt.show()
```

```python
correlation_coefficient = df['Infrastructure'].corr(df['Total score'])
print(f"Correlation Coefficient: {correlation_coefficient}")  # Positive Correlation but not stronger than the Talent correlation coefficient.
```

```python
plt.scatter(df['Operating Environment'], df['Total score'])
plt.title("Correlation of Operating Environment and Total Score")
plt.xlabel("Operating Environment")
plt.ylabel("Total Score") # Weaker positive correlation: The more the regulatory context, and public opinion surrounding artificial intelligence, the higher the total score in any country.

plt.tight_layout()
plt.show()
```

```python
correlation_coefficient = df['Operating Environment'].corr(df['Total score'])  # Weak to no Positive Correlation.
print(f"Correlation Coefficient: {correlation_coefficient}")
```

```python
# Among the AI Implementation Factors, Talent has the most impact on the Total Score. More the human capital in a country, the higher its long-run economic growth.
```

```python
plt.scatter(df['Research'], df['Total score'])
plt.title("Correlation of Research and Total Score")
plt.xlabel("Research")   # Strong Positive Correlation: The more the extent of specialist research and researchers; investigating the amount of publications and citations in credible academic journals, the higher the total score.
plt.ylabel("Total Score") # Hence, the more research on the potential for artificial intelligence to evolve and improve, the higher the total score.

plt.tight_layout()
plt.show()
```

```python
correlation_coefficient = df['Research'].corr(df['Total score'])  # The research correlation coefficient indicates that the relationship is a solid positive one.
print(f"Correlation Coefficient: {correlation_coefficient}")
```

```python
plt.scatter(df['Development'], df['Total score'])
plt.title("Correlation of Development and Total Score")
plt.xlabel("Development")  # Positive Correlation: The more the development of fundamental platforms and algorithms upon which innovative artificial intelligence projects rely, the higher the total score.
plt.ylabel("Total Score")

plt.tight_layout()
plt.show()
```

```python
correlation_coefficient = df['Development'].corr(df['Total score']) # The development correlation coefficient indicates that the relationship is also a strong positive one.
print(f"Correlation Coefficient: {correlation_coefficient}") # However, Research has a more powerful influence on the total score.
```

```python
# Between the two innovation factors, Research has the most impact on the Total Score because it can improve the current AI technologies that can help a country achieve long-run economic growth.
```

```python
plt.scatter(df['Government Strategy'], df['Total score'])
plt.title("Correlation of Government Strategy and Total Score")
plt.xlabel("Government Strategy")  # Weak Positive Correlation: The more the depth of commitment from the national government to artificial intelligence; investigating spending commitments and national strategies, the higher the total score.
plt.ylabel("Total Score")

plt.tight_layout()
plt.show()
```

```python
correlation_coefficient = df['Government Strategy'].corr(df['Total score'])  # Weak Positive Correlation.
print(f"Correlation Coefficient: {correlation_coefficient}")
```

```python
plt.scatter(df['Commercial'], df['Total score'])
plt.title("Correlation of Commercial and Total Score")
plt.xlabel("Commercial")  # Strong Positive Correlation: The more the level of startup activity, investment, and business initiatives based on artificial intelligence, the higher the total score.
plt.ylabel("Total Score")

plt.tight_layout()
plt.show()
```

```python
correlation_coefficient = df['Commercial'].corr(df['Total score'])  # Very Strong Positive Correlation.
print(f"Correlation Coefficient: {correlation_coefficient}")
```

```python
# Between the two investment factors, Commercial has the most impact on the Total Score because it is directly linked with investment and business confidence, both of which increase the aggregate demand and help a country achieve short-run economic growth.
```
